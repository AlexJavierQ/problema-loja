# Hallazgos · demostración en el navegador
Ejecuta las tres celdas en orden en un entorno de CPU. La primera ejecución descarga los modelos. Los participantes de una demostración pueden representarse con dos reportes en el mismo navegador. Cada sesión de Colab tiene su propia base temporal. Usa objetos de muestra.

```
from pathlib import Path
import subprocess, sys
raiz = Path('/content/hallazgos-loja')
if not raiz.exists():
    subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/AlexJavierQ/problema-loja.git', str(raiz)], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', str(raiz/'requirements.txt')], check=True)
subprocess.run([sys.executable, str(raiz/'hallazgos/modelos.py')], check=True)
```

```
import time, urllib.request
if 'servidor' not in globals() or servidor.poll() is not None:
    registro = open('/tmp/hallazgos.log', 'w')
    servidor = subprocess.Popen([sys.executable, str(raiz/'hallazgos/servidor.py')], cwd=raiz, stdout=registro, stderr=registro)
for intento in range(90):
    if servidor.poll() is not None:
        raise RuntimeError(Path('/tmp/hallazgos.log').read_text()[-2000:])
    try:
        with urllib.request.urlopen('http://127.0.0.1:5180/', timeout=2): break
    except OSError:
        time.sleep(1)
else:
    raise TimeoutError('El servidor tarda en iniciar. Revisa /tmp/hallazgos.log.')
print('Hallazgos está listo.')
```
